# AAVAIL Revenue Forecasting EDA
This notebook demonstrates the visualization stage of the AI workflow.

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv('../data/monthly_revenue.csv', parse_dates=['date'])
pivot = df.pivot(index='date', columns='country', values='revenue')
ax = pivot.plot(figsize=(12,6), title='AAVAIL Monthly Revenue by Country')
ax.set_ylabel('Revenue')
ax.grid(alpha=0.25)
plt.show()
```
